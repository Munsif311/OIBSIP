```python
df["date"] = pd.to_datetime(df["date"])
```

```python
monthly_sales = (
    df.groupby(df["date"].dt.to_period("M"))["revenue"]
      .sum()
)
```

```python
monthly_sales.plot(figsize=(14,6), marker="o")
plt.title("Monthly Revenue")
plt.show()
```

```python
quarterly_sales = (
    df.groupby(df["date"].dt.to_period("Q"))["revenue"]
      .sum()
)
quarterly_sales
```

```python
top_products = (
    df.groupby("product_id")["revenue"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)
top_products
```

```python
top_sales = (
    df.groupby("product_id")["sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)
top_sales
```

```python
top_store = (
    df.groupby("store_id")["revenue"]
      .sum()
      .sort_values(ascending=False)
)
top_store
```

```python
df.groupby("promo_type_1")["revenue"].mean()
```
